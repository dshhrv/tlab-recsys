# GPT-2: next-item prediction
Обучение с нуля на train; validation GTS Last, test GTS Random. Полный активный train-каталог без добавления target.


```python
from dataset import device_for_run, load_store
from gpt2 import GPTSettings, train_gpt
from pathlib import Path 
from evaluate import evaluate_variant
from tracking import start_task
from dataclasses import asdict
```

```python
ROOT = Path.cwd()
DATA = ROOT / 'data' / 'Yambda'
CHECKPOINT = ROOT / 'checkpoints' / 'gpt2'
RESULT = ROOT / 'results' / 'gpt2_metrics.json'
LOG_CLEARML = True
EVALUATE_TEST = True
settings = GPTSettings(max_items=30000, max_length=128, embedding_dim=128, layers=2, heads=4, dropout=0.1, epochs=15, batch_size=16, learning_rate=3e-4, weight_decay=0.01, gradient_clip=1.0, max_windows=None, validation_queries=2000, seed=42)
device = device_for_run()
store = load_store(DATA, settings.max_items, settings.max_length)
print('device:', device, 'active catalog:', len(store.catalog.items), 'train catalog:', store.info['train_catalog_items'])
```

```python
task = start_task('GPT-2 baseline', {'settings': asdict(settings), 'catalog_items': len(store.catalog.items)}, LOG_CLEARML)
model, history = train_gpt(store, settings, CHECKPOINT, device, task)
results = evaluate_variant(store, model, model, None, None, RESULT, device, task=task, include_test=EVALUATE_TEST)
if task:
    task.close()
print(history)
print(results['validation']['summary'])
if EVALUATE_TEST:
    print(results['test']['summary'])
```
